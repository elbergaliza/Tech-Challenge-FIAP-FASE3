```python
import os

# Modelo fine-tuned publicado no HuggingFace Hub
# (treinado via tunning/05_treino_completo_colab.ipynb)
os.environ["HF_PIPELINE_MODEL"] = "thallesf1/qwen2.5-3b-medpt-lora"

from carregador_do_modelo import carregar_modelo
from banco_de_conhecimento import construir_indice
from fluxo import montar_fluxo
from configs import CONFIG_DENGUE, CONFIG_COVID
from nos.confirmacao import criar_no_confirmacao
from nos.tratamento import criar_no_tratamento

modelo = carregar_modelo()
indice = construir_indice([CONFIG_DENGUE, CONFIG_COVID])
fluxo  = montar_fluxo(
    modelo,
    indice,
    incluir_confirmacao_tratamento=False,
    configs=[CONFIG_DENGUE, CONFIG_COVID]
)
no_confirmacao = criar_no_confirmacao()
no_tratamento  = criar_no_tratamento(indice, modelo)

print("Sistema pronto!")
```

```python
from main import fazer_triagem_inicial, aplicar_confirmacao_e_tratamento,_exibir_resultado_triagem, _exibir_resultado_final

# Descreve os sintomas do paciente
sintomas = "paciente com febre alta, dor de cabeça intensa e manchas vermelhas pelo corpo"

# Executa a triagem
resultado = fazer_triagem_inicial(sintomas, fluxo)

# Exibe o resultado formatado
_exibir_resultado_triagem(resultado)
```

```python
resultado_final = aplicar_confirmacao_e_tratamento(
    resultado,
    {"human_in_the_loop": True},
    no_confirmacao,
    no_tratamento,
)

_exibir_resultado_final(resultado_final)
```
